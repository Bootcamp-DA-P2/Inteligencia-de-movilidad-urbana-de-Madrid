### Paso 2: Transformaciones

#### Creación de una columna de tipo fecha y hora a partir del id

```python
df = (
    df.with_columns(
        pl.datetime(
            year=pl.col("id_fecha").dt.year(),
            month=pl.col("id_fecha").dt.month(),
            day=pl.col("id_fecha").dt.day(),
            hour=pl.col("hora"),
            minute=pl.lit(0),
            second=pl.lit(0),
        ).alias("fecha_hora")
    )
    .sort(["id_sensor", "fecha_hora"])
)
```

#### Visualización de las columnas de fecha generadas

```python
df.select(
    [
        "id_sensor",
        "id_fecha",
        "hora",
        "fecha_hora",
    ]
).head(10)
```

#### Visualización de la Agrupación por Sensor

```python
(
    df
    .group_by("id_sensor")
    .agg(
        [
            pl.col("fecha_hora").min().alias("inicio"),
            pl.col("fecha_hora").max().alias("fin"),
            pl.len().alias("registros"),
        ]
    )
    .head()
)
```

#### Calculo dse del tiempo transcurrido entre una medición y la anterior

```python
df = df.with_columns(
    pl.col("fecha_hora")
    .diff()
    .over("id_sensor")
    .alias("delta_tiempo")
)
```

#### Visualización de la frecuencia de los intervalos de tiempo

```python
(
    df
    .group_by("delta_tiempo")
    .len()
    .sort("delta_tiempo")
)
```

#### Detección de todos los registros en los que existe un salto temporal entre mediciones

```python
huecos = (
    df
    .filter(
        (pl.col("delta_tiempo").is_not_null()) &
        (pl.col("delta_tiempo") != pl.duration(hours=1))
    )
)

print(f"Registros con saltos temporales: {huecos.height:,}")
```

#### Visualización de los huecos detectados

```python
huecos.select(
    [
        "id_sensor",
        "fecha_hora",
        "delta_tiempo",
    ]
).head(20)
```

#### Eliminación de la columna `delta_tiempo`

```python
df = df.drop("delta_tiempo")
```

#### Visualización de valores duplicados

```python
duplicados = (
    df
    .group_by(["id_sensor", "fecha_hora"])
    .len()
    .filter(pl.col("len") > 1)
)

print(f"Duplicados encontrados: {duplicados.height}")
```

### Paso 3: Feature Engineering

#### Transformación de las variables temporales en variables cíclicas usando las funciones seno y coseno

```python
# Transformación cíclica de variables temporales

df = df.with_columns([
    (2 * np.pi * pl.col("hora") / 24).sin().alias("hora_sin"),
    (2 * np.pi * pl.col("hora") / 24).cos().alias("hora_cos"),

    (2 * np.pi * pl.col("dia_semana") / 7).sin().alias("dia_semana_sin"),
    (2 * np.pi * pl.col("dia_semana") / 7).cos().alias("dia_semana_cos"),

    (2 * np.pi * pl.col("mes") / 12).sin().alias("mes_sin"),
    (2 * np.pi * pl.col("mes") / 12).cos().alias("mes_cos"),
])
```

#### Visualización de las tranformaciones realizadas

```python
df.select([
    "hora",
    "hora_sin",
    "hora_cos"
]).head()
```

#### definición de las variables del dataset que van a utilizarse para crear lag features

```python
variables_lag = [
    "intensidad_media",
    "ocupacion_media",
    "velocidad_media"
]

lags = [1, 24, 168]
```

#### Creación de las lag_features

```python
lag_exprs = []

for variable in variables_lag:
    for lag in lags:
        lag_exprs.append(
            pl.col(variable)
            .shift(lag)
            .over("id_sensor")
            .alias(f"{variable}_lag_{lag}")
        )

df = df.with_columns(lag_exprs)
```

#### Visualización de las lag_features

```python
df.select([
    "ocupacion_media",
    "ocupacion_media_lag_1",
    "ocupacion_media_lag_24",
    "ocupacion_media_lag_168"
]).head(10)
```

#### Creación de las columnas con estadísticas móviles

```python
rolling_exprs = [

    # Intensidad
    pl.col("intensidad_media")
    .rolling_mean(window_size=3)
    .over("id_sensor")
    .alias("rolling_intensidad_mean_3"),

    pl.col("intensidad_media")
    .rolling_mean(window_size=24)
    .over("id_sensor")
    .alias("rolling_intensidad_mean_24"),

    pl.col("intensidad_media")
    .rolling_std(window_size=24)
    .over("id_sensor")
    .alias("rolling_intensidad_std_24"),


    # Ocupación
    pl.col("ocupacion_media")
    .rolling_mean(window_size=3)
    .over("id_sensor")
    .alias("rolling_ocupacion_mean_3"),

    pl.col("ocupacion_media")
    .rolling_mean(window_size=24)
    .over("id_sensor")
    .alias("rolling_ocupacion_mean_24"),

    pl.col("ocupacion_media")
    .rolling_std(window_size=24)
    .over("id_sensor")
    .alias("rolling_ocupacion_std_24"),

]

df = df.with_columns(rolling_exprs)
```

#### Visualización de las variables creadas sobre las estadísticas móviles

```python
df.select([
    "ocupacion_media",
    "rolling_ocupacion_mean_3",
    "rolling_ocupacion_mean_24",
    "rolling_ocupacion_std_24"
]).head(10)
```

#### Cálculo de diferencias puntuales respecto a períodos anteriores

```python
df = df.with_columns([

    # Intensidad
    (
        pl.col("intensidad_media")
        - pl.col("intensidad_media_lag_1")
    ).alias("delta_intensidad_1"),

    (
        pl.col("intensidad_media")
        - pl.col("intensidad_media_lag_24")
    ).alias("delta_intensidad_24"),

    # Ocupación
    (
        pl.col("ocupacion_media")
        - pl.col("ocupacion_media_lag_1")
    ).alias("delta_ocupacion_1"),

    (
        pl.col("ocupacion_media")
        - pl.col("ocupacion_media_lag_24")
    ).alias("delta_ocupacion_24"),

])
```

#### Visualización de las variables creadas anteriormente

```python
df.select([
    "ocupacion_media",
    "ocupacion_media_lag_1",
    "delta_ocupacion_1",
    "ocupacion_media_lag_24",
    "delta_ocupacion_24"
]).head(10)
```

#### Creación de la variable target para el entrenamiento del modelo de Machine Learning

```python
df = df.with_columns(
    pl.col("ocupacion_media")
    .shift(-1)
    .over("id_sensor")
    .alias("target")
)
```

#### Visualizaci´n de la variable target

```python
df.select([
    "fecha_hora",
    "ocupacion_media",
    "target"
]).head(10)
```

#### Filtrado de targets inválidos por saltos temporales

El `shift(-1)` asume que la fila siguiente de cada sensor corresponde exactamente a la hora siguiente. Sin embargo, existen 132.328 saltos temporales (huecos de 2h a 173 días). En esas filas el target no es la ocupación de t+1 sino la de un instante muy posterior — etiquetas incorrectas que contaminan el entrenamiento. Se filtran las filas cuyo registro siguiente no está exactamente a 1 hora.

```python
filas_antes_filtro = df.height

df = df.with_columns(
    pl.col("fecha_hora")
    .shift(-1)
    .over("id_sensor")
    .alias("fecha_hora_siguiente")
)

df = df.filter(
    (pl.col("fecha_hora_siguiente") - pl.col("fecha_hora"))
    == pl.duration(hours=1)
).drop("fecha_hora_siguiente")

filas_despues_filtro = df.height

print(f"Filas antes del filtro : {filas_antes_filtro:,}")
print(f"Filas después del filtro: {filas_despues_filtro:,}")
print(f"Targets inválidos eliminados: {filas_antes_filtro - filas_despues_filtro:,}")
```
