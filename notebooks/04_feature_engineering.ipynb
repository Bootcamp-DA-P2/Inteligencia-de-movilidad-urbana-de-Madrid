```python
import polars as pl
import numpy as np
```

```python
# Cargar datos preparados tras el proceso de limpieza y EDA
df = pl.read_parquet("../data/modeling/base_modelado.parquet")

print(f"Filas: {df.height:,}")
print(f"Columnas: {df.width}")

df.head()
```

```python
df.schema
```

```python
df.describe()
```

```python
df = (
    df.with_columns(
        pl.datetime(
            year=pl.col("id_fecha").dt.year(),
            month=pl.col("id_fecha").dt.month(),
            day=pl.col("id_fecha").dt.day(),
            hour=pl.col("hora"),
            minute=pl.lit(0),
            second=pl.lit(0),
        ).alias("fecha_hora")
    )
    .sort(["id_sensor", "fecha_hora"])
)
```

```python
df.select(
    [
        "id_sensor",
        "id_fecha",
        "hora",
        "fecha_hora",
    ]
).head(10)
```

```python
(
    df
    .group_by("id_sensor")
    .agg(
        [
            pl.col("fecha_hora").min().alias("inicio"),
            pl.col("fecha_hora").max().alias("fin"),
            pl.len().alias("registros"),
        ]
    )
    .head()
)
```

```python
df = df.with_columns(
    pl.col("fecha_hora")
    .diff()
    .over("id_sensor")
    .alias("delta_tiempo")
)
```

```python
(
    df
    .group_by("delta_tiempo")
    .len()
    .sort("delta_tiempo")
)
```

```python
huecos = (
    df
    .filter(
        (pl.col("delta_tiempo").is_not_null()) &
        (pl.col("delta_tiempo") != pl.duration(hours=1))
    )
)

print(f"Registros con saltos temporales: {huecos.height:,}")
```

```python
huecos.select(
    [
        "id_sensor",
        "fecha_hora",
        "delta_tiempo",
    ]
).head(20)
```

```python
df = df.drop("delta_tiempo")
```

```python
duplicados = (
    df
    .group_by(["id_sensor", "fecha_hora"])
    .len()
    .filter(pl.col("len") > 1)
)

print(f"Duplicados encontrados: {duplicados.height}")
```

```python
print("=" * 50)
print("PREPARACIÓN COMPLETADA")
print("=" * 50)

print(f"Filas: {df.height:,}")
print(f"Columnas: {df.width:,}")
print(f"Sensores únicos: {df['id_sensor'].n_unique():,}")
```

```python
# Transformación cíclica de variables temporales

df = df.with_columns([
    (2 * np.pi * pl.col("hora") / 24).sin().alias("hora_sin"),
    (2 * np.pi * pl.col("hora") / 24).cos().alias("hora_cos"),

    (2 * np.pi * pl.col("dia_semana") / 7).sin().alias("dia_semana_sin"),
    (2 * np.pi * pl.col("dia_semana") / 7).cos().alias("dia_semana_cos"),

    (2 * np.pi * pl.col("mes") / 12).sin().alias("mes_sin"),
    (2 * np.pi * pl.col("mes") / 12).cos().alias("mes_cos"),
])
```

```python
df.select([
    "hora",
    "hora_sin",
    "hora_cos"
]).head()
```

```python
variables_lag = [
    "intensidad_media",
    "ocupacion_media",
    "velocidad_media"
]

lags = [1, 24, 168]
```

```python
lag_exprs = []

for variable in variables_lag:
    for lag in lags:
        lag_exprs.append(
            pl.col(variable)
            .shift(lag)
            .over("id_sensor")
            .alias(f"{variable}_lag_{lag}")
        )

df = df.with_columns(lag_exprs)
```

```python
df.select([
    "ocupacion_media",
    "ocupacion_media_lag_1",
    "ocupacion_media_lag_24",
    "ocupacion_media_lag_168"
]).head(10)
```

```python
rolling_exprs = [

    # Intensidad
    pl.col("intensidad_media")
    .rolling_mean(window_size=3)
    .over("id_sensor")
    .alias("rolling_intensidad_mean_3"),

    pl.col("intensidad_media")
    .rolling_mean(window_size=24)
    .over("id_sensor")
    .alias("rolling_intensidad_mean_24"),

    pl.col("intensidad_media")
    .rolling_std(window_size=24)
    .over("id_sensor")
    .alias("rolling_intensidad_std_24"),


    # Ocupación
    pl.col("ocupacion_media")
    .rolling_mean(window_size=3)
    .over("id_sensor")
    .alias("rolling_ocupacion_mean_3"),

    pl.col("ocupacion_media")
    .rolling_mean(window_size=24)
    .over("id_sensor")
    .alias("rolling_ocupacion_mean_24"),

    pl.col("ocupacion_media")
    .rolling_std(window_size=24)
    .over("id_sensor")
    .alias("rolling_ocupacion_std_24"),

]

df = df.with_columns(rolling_exprs)
```

```python
df.select([
    "ocupacion_media",
    "rolling_ocupacion_mean_3",
    "rolling_ocupacion_mean_24",
    "rolling_ocupacion_std_24"
]).head(10)
```

```python
df = df.with_columns([

    # Intensidad
    (
        pl.col("intensidad_media")
        - pl.col("intensidad_media_lag_1")
    ).alias("delta_1"),

    (
        pl.col("intensidad_media")
        - pl.col("intensidad_media_lag_24")
    ).alias("delta_24"),

    # Ocupación
    (
        pl.col("ocupacion_media")
        - pl.col("ocupacion_media_lag_1")
    ).alias("delta_ocupacion_1"),

    (
        pl.col("ocupacion_media")
        - pl.col("ocupacion_media_lag_24")
    ).alias("delta_ocupacion_24"),

])
```

```python
df.select([
    "ocupacion_media",
    "ocupacion_media_lag_1",
    "delta_ocupacion_1",
    "ocupacion_media_lag_24",
    "delta_ocupacion_24"
]).head(10)
```

```python
df = df.with_columns(
    pl.col("ocupacion_media")
    .shift(-1)
    .over("id_sensor")
    .alias("target")
)
```

```python
df.select([
    "fecha_hora",
    "ocupacion_media",
    "target"
]).head(10)
```

```python
(
    df
    .null_count()
    .transpose(include_header=True)
    .filter(pl.col("column_0") > 0)
)
```

```python
df = df.drop([
    "id_fecha",
    "nombre_mes"
])
```
